# Chi-Squared Test

Basic Idea:<br>
Compare the observed frequencies in your data with the frequencies you would expect if the null hypothesis were true.

The Formula is: <h3>$\chi^2 = \sum \frac{(O-E)^2}{E}$</h2> <br>
Where <br>$O$ = observed frequency <br>
$E$ = expected frequency<br>

Wherea a large $\chi^2$ value means the observed data differ strongly from what was expected.

## 1. Chi-square Goodness-of-Fit Test



The **Chi-square goodness-of-fit test** is used to determine whether the observed frequencies of one categorical variable differ significantly from an expected distribution.

The hypotheses are:

$$
H_0: \text{The observed distribution follows the expected distribution}
$$

$$
H_1: \text{The observed distribution does not follow the expected distribution}
$$

The Chi-square test statistic is:

$$
\chi^2 = \sum \frac{(O_i-E_i)^2}{E_i}
$$

where:

- $O_i$ = observed frequency of category $i$
- $E_i$ = expected frequency of category $i$

The degrees of freedom are:

$$
df = k - 1
$$

where $k$ is the number of categories.

The decision rule is:

$$
p < \alpha \Rightarrow \text{reject } H_0
$$

$$
p \geq \alpha \Rightarrow \text{do not reject } H_0
$$

### Example:

We ask 100 students what mode of transport they use.
The results are the following:
| Transportation mode | Number of students |
|---|---:|
| Car | 35 |
| Public transport | 25 |
| Bicycle | 20 |
| Walking | 20 |


We assume that all four modes of transportation are equally preferred.

Using a significance level of $\alpha = 0.05$ (or 5%), test whether the observed distribution of transportation preferences differs significantly from an equal distribution.

```python
# Python code: 

from scipy.stats import chisquare

# Observed frequencies
observed = [35, 25, 20, 20]

# Expected frequencies - for all are equally preferred
# (total students / number of categories) -> 100/4 = 25
expected = [25, 25, 25, 25]

chi2, p_value = chisquare(
    f_obs=observed,
    f_exp=expected
)

print("Chi-square:", chi2)
print("p-value:", p_value)
```

```python
# R Code:

# Observed frequencies
observed <- c(35, 25, 20, 20)

# Expected probabilities
# All four categories are equally preferred
expected_prob <- c(0.25, 0.25, 0.25, 0.25)

result <- chisq.test(
  x = observed,
  p = expected_prob
)

print(result)
```

#### Calculation:

To calculate this we need to compare the observed values with the expected values under our $H_0$.<br>
In our example we expect an even distribution, meaning that all transportation modes are used the same amount.<br>
We can split 100 evenly in 4 categories, so we expect each to be 25.<br>

| Transportation mode | Number of students observed | Number of students expected
|---|---:|--:|
| Car | 35 | 25 |
| Public transport | 25 | 25 |
| Bicycle | 20 | 25 | 
| Walking | 20 | 25 |

We can now use the Chi-square statistic ($\chi^2 = \sum \frac{(O_i-E_i)^2}{E_i}$) to calculate $\chi^2$.

$$
\chi^2 = \frac{(35-25)^2}{25} + \frac{(25-25)^2}{25} + [...]
$$

From this we get a $\chi^2$ of 6. <br><br>
And our degrees of freedom ( $df = k - 1$ ) is 3.



### Interpretation:
The p-value is approximately:

$$
p \approx 0.11
$$

The significance level is:

$$
\alpha = 0.05
$$

We compare the p-value with the significance level:

$$
0.11 > 0.05
$$

Therefore, we **do not reject $H_0$**.

Thus we conclude:
$$
\text{There is not enough evidence to say that the transportation preferences differ significantly from an equal distribution.}
$$

## 2. Chi-square Test of Independence

The **Chi-square test of independence** is used to determine whether there is a statistically significant association between **two categorical variables**.

The hypotheses are:
$$
H_0: \text{The two categorical variables are independent.}
$$

$$
H_1: \text{The two categorical variables are associated.}
$$
The expected frequency for each cell is calculated as:

$$
E_{ij} =
\frac{
(\text{row total}_i)(\text{column total}_j)
}{
\text{grand total}
}
$$

The Chi-square statistic is:

$$
\chi^2 =
\sum
\frac{(O_{ij}-E_{ij})^2}{E_{ij}}
$$

where:

- $O_{ij}$ = observed frequency
- $E_{ij}$ = expected frequency

The degrees of freedom are calculated as:

$$
df = (r-1)(c-1)
$$

where:

- $r$ = number of rows
- $c$ = number of columns

The decision rule is:

If $p < \alpha$, reject $H_0$.

If $p \geq \alpha$, do not reject $H_0$.


### Example



Say a university wants to investigate whether **transportation preference is associated with the subject they are studying**.

The following data were collected amongst students studying Computer Science (Group A) and students studying Environmental Science (Group B):

| Student Group | Car | Public Transport | Bicycle |
|---|---:|---:|---:|
| Group A | 120 | 140 | 30 |
| Group B | 30 | 20 | 10 |

The two categorical variables are:

- **Student group:** Group A / Group B
- **Transportation mode:** Car / Public Transport / Bicycle

Because we want to determine whether two categorical variables are associated, we use a **Chi-square test of independence**.

Using a significance level of $\alpha = 0.05$ (or 5%), test whether transportation preference is associated with student group.

### Hypotheses

$H_0$: Transportation preference and student group are independent.

$H_1$: Transportation preference and student group are associated.

```python
# Python code:

import pandas as pd
from scipy.stats import chi2_contingency

# Create contingency table
observed = pd.DataFrame(
    [
        [120, 140, 30],
        [30, 20, 10]
    ],
    index=["Group A", "Group B"],
    columns=["Car", "Public Transport", "Bicycle"]
)

# Display observed frequencies
display(observed)

# Perform Chi-square test of independence
chi2, p_value, df, expected = chi2_contingency(observed)

print("Chi-square:", chi2)
print("Degrees of freedom:", df)
print("p-value:", p_value)

# Display expected frequencies with labels
expected = pd.DataFrame(
    expected,
    index=observed.index,
    columns=observed.columns
)

display(expected)
```

```python
# R code:

# Create contingency table
observed <- matrix(
  c(
    120, 140, 30,
    30, 20, 10
  ),
  nrow = 2,
  byrow = TRUE
)

# Add names
rownames(observed) <- c("Group A", "Group B")
colnames(observed) <- c("Car", "Public Transport", "Bicycle")

print(observed)

# Perform Chi-square test of independence
result <- chisq.test(observed)

# Print test result
print(result)

# Print expected frequencies
print(result$expected)
```

### Calculation:

To help calculate the expected frequencies manually we can add the column total and row total to our table, the cell on the bottom right of the table is the grand total (The total number of students surveyed).

| Student Group | Car | Public Transport | Bicycle | row total |
|---|---:|---:|---:|---:|
| Group A | 120 | 140 | 30 | 290 |
| Group B | 30 | 20 | 10 | 60 |
| column total | 150 | 160 | 40 | 350 |

For each cell:

$$
E_{ij} =
\frac{
(\text{row total})(\text{column total})
}{
\text{grand total}
}
$$

For example, the expected number of students in **Group A who use a car** is:

$$
E =
\frac{290 \cdot 150}{350}
$$

$$
E \approx 125
$$

Doing this for every entry in the complete expected-frequency table results in this table:

| Student Group | Car | Public Transport | Bicycle |
|---|---:|---:|---:|
| Group A | ~125 | ~132 | ~33 |
| Group B | ~25 | ~27 | ~7 |

We can now use the Chi-square statistic ($\chi^2 = \sum \frac{(O_i-E_i)^2}{E_i}$) to compare the tables.

$$
\chi^2 = \frac{(120-125)^2}{125} + \frac{(140-132)^2}{132} + [...]
$$

From this we get a $\chi^2$ of ~5. 
And our degree of freedom ( $df = (r-1)(c-1)$ ) are :

$$
df = (2-1)(3-1)
$$
$$
df = 2
$$




### Interpretation:

The p-value is approximately:

$$
p \approx 0.08
$$

The significance level is:

$$
\alpha = 0.05
$$

We compare the p-value with the significance level:

$$
0.08 > 0.05
$$

Therefore, we **do not reject $H_0$**.

Thus we conclude:
$$
\text{The Chi-square test of independence showed no statistically significant association between student group and transportation preference.}
$$
